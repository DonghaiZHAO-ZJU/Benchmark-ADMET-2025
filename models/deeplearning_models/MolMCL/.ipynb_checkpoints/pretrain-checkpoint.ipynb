```python
smiles = pd.read_csv('~/project/RobustAC/data/pretrain/zinc_smiles.csv')['smiles']
```

```python
fr_SMARTS = {}
fr_PATT = {}
with open(os.path.join(RDConfig.RDDataDir, 'FragmentDescriptors.csv'), 'r') as f:
    for line in f.readlines():
        if len(line) and line[0] != '#':
            splitL = line.rstrip().split('\t')
            if len(splitL) >= 3:
                name, descr, sma = splitL[:3]
                name = name.replace('=', '_')
                name = name.replace('-', '_')
                descr = descr.replace('"', '')
                patt = Chem.MolFromSmarts(sma)
                if patt is None or patt.GetNumAtoms() == 0:
                    print('Smarts %s could not be parsed' % (repr(sma)))
                    raise ImportError
                fr_SMARTS[name] = sma
                fr_PATT[name] = patt
fr_FUNCS = {name: func for name, func in Descriptors.descList if name in fr_SMARTS}


def generate_motif_features(mol, binary=True, verbose=False):
    features = np.zeros(len(fr_FUNCS), dtype=int)
    for i, fr in enumerate(fr_FUNCS):
        features[i] = fr_FUNCS[fr](mol)
        if binary:
            features[i] = int(features[i] > 0)
            if features[i] > 0 and verbose:
                print('Match {}: {}'.format(fr, fr_SMARTS[fr]))
            elif verbose:
                print('Unmatch {}'.format(fr))
    return features
```

```python
def get_reps_local(model, data):
    model.eval()
    with torch.no_grad():
        _, node_repres = model.gnn(data.x, data.edge_index, data.edge_attr, data.batch) 
        h_g_0, _, _ = model.aggrs[0](node_repres, data.batch)
        h_g_1, _, _ = model.aggrs[1](node_repres, data.batch)
        h_g_2, _, _ = model.aggrs[2](node_repres, data.batch)
    h_g_0 = F.normalize(h_g_0, dim=-1).view(-1)
    h_g_1 = F.normalize(h_g_1, dim=-1).view(-1)
    h_g_2 = h_g_2.view(-1)
    return h_g_0, h_g_1, h_g_2

def tanimoto(v1, v2):
    return(np.bitwise_and(v1, v2).sum() / np.bitwise_or(v1, v2).sum())

mol_sim_list, scaff_sim_list, motif_sim_list = [], [], []
mcd_dist_list, scd_dist_list, cp_dist_list = [], [], []

np.random.seed(2)
rand_pair_inds = np.array([np.random.randint(len(smiles), size=2) for _ in range(1000)])
for idx1, idx2 in tqdm(rand_pair_inds):
    smi1, smi2 = smiles[idx1], smiles[idx2]
    mol1, mol2 = Chem.MolFromSmiles(smi1), Chem.MolFromSmiles(smi2)
    scaff1, scaff2 = get_scaffold(smi1), get_scaffold(smi2)
    
    mol_fp1, mol_fp2 = get_fp(Chem.MolFromSmiles(smi1)), get_fp(Chem.MolFromSmiles(smi2))
    scaff_fp1, scaff_fp2 = get_fp(Chem.MolFromSmiles(scaff1)), get_fp(Chem.MolFromSmiles(scaff2))
    motif_feat1, motif_feat2 = generate_motif_features(mol1), generate_motif_features(mol2)
    
    mol_sim = DataStructs.BulkTanimotoSimilarity(mol_fp1, [mol_fp2])[0]
    scaff_sim = DataStructs.BulkTanimotoSimilarity(scaff_fp1, [scaff_fp2])[0]
    motif_sim = tanimoto(motif_feat1, motif_feat2)
    
    data1, data2 = mol_to_graph_data_obj_super_rich(mol1), mol_to_graph_data_obj_super_rich(mol2)
    
    h1_mcd, h1_scd, h1_cp = get_reps_local(model, data1)
    h2_mcd, h2_scd, h2_cp = get_reps_local(model, data2)
    
    mol_sim_list.append(mol_sim)
    scaff_sim_list.append(scaff_sim)
    motif_sim_list.append(motif_sim)
    
    mcd_dist_list.append((h1_mcd - h2_mcd).norm(dim=-1).item()) 
    scd_dist_list.append((h1_scd - h2_scd).norm(dim=-1).item()) 
    cp_dist_list.append((h1_cp - h2_cp).norm(dim=-1).item())  
    break
```
