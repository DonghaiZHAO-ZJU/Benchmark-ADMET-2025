```python
import torch
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm
from matplotlib import colormaps
```

## Pretrain Ablation:

```python
results = pd.read_csv('pretrain_ablation.csv')
for name in results.columns[2:]:
    stats = [stat.split(' ') for stat in results[name]]
    mu = [float(stat[0]) for stat in stats]
    std = [float(stat[1][1:-1]) for stat in stats]
    print(f'Average R2 of {name} = {np.mean(mu)}')

original_name = 'ab0(paper)'
mu = results[original_name]
print(f'Average R2 of {original_name} = {np.mean(mu)}')
```

```python
results = pd.read_csv('pretrain_ablation_wln.csv')
scores, scores_std = [], []
scores_flatten = []
for name in results.columns[4:]: 
    stats = [stat.split(' ') for stat in results[name]]
    mu = [float(stat[0]) for stat in stats]
    std = [float(stat[1][1:-1]) for stat in stats]
    name = name.replace('\n', ' ')
    scores.append(np.mean(mu))
    scores_std.append(np.mean(std))
    scores_flatten.append(mu)
    print(f'Average R2 of {name} = {np.mean(mu)}')

original_name = 'ab0(paper)'
mu = results[original_name]
scores_flatten.append(mu)
scores.append(np.mean(mu))
scores_std.append(0.016164862487057126)
print(f'Average R2 of {original_name} = {np.mean(mu)}')
```

```python
scores_flatten
```

```python
plt.figure(figsize=(12, 4))
methods = ['Full',
           'No regularizations', 
           'No intra-channel regularization', 
           'No inter-channel regularization',
           'Replace with margin loss', 
           'Replace with multi-task learning']

selected_idx = [0, 1, 2, 3, 4, 5]
selected_scores = np.array([scores[i] for i in selected_idx])
values = selected_scores# - scores[0]
selected_stds = np.array([scores_std[i] for i in selected_idx])
selected_methods = [methods[i] for i in selected_idx]

plt.barh(selected_methods[::-1], values[::-1], 
         xerr=selected_stds, capsize=10, color=colormaps['tab20b'](np.arange(len(values))))
plt.axvline(x=0, color='red', linestyle='--', linewidth=2)

# # Overlay dot plots for each category
# for i, category in enumerate(methods):
#     jitter = np.random.normal(0, 0.1, size=len(scores_flatten[i]))
#     plt.scatter(scores_flatten[i], np.arange(len(selected_idx))[i] + jitter, color='darkblue', s=50, alpha=0.7)

ax = plt.gca()  # Get current axis
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['bottom'].set_visible(True)

plt.xlabel('Average Test R-squared', fontsize=16)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.title('Ablation in Pre-train Components', fontsize=16)
plt.xlim(0.55, 0.65)
plt.savefig('./figure/pretrain_ablation.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Finetune Ablation:

Takeaway:

- Concat works slightly worse than Aggr
- MoE

```python
results = pd.read_csv('finetune_ablation.csv')
```

```python
scores, scores_std = [], []
for i, name in enumerate(results.columns[3:]):
    stats = [stat.split(' ') for stat in results[name]]
    mu = [float(stat[0]) for stat in stats]
    std = [float(stat[1][1:-1]) for stat in stats]
    name = name.replace('\n', ' ')
    print(f'{i}, Average R2 of {name} = {np.mean(mu)}')
    scores.append(np.mean(mu))
    scores_std.append(np.std(std))

original_name = 'gin (paper)'
mu = results[original_name]
scores.append(np.mean(mu))
scores_std.append(0.016164862487057126)
print(f'Average R2 of {original_name} = {np.mean(mu)}')
```

```python
plt.figure(figsize=(12, 4))
methods = ['',
           'PW', 
           '*Sparse PW (t=0.3)', 
           '',
           'Concat', 
           '',
           'MoE',
           '',
           'Sparse MoE (t=0.7)',
           '',
           '',
           '*Sparse MoE (t=0.3)',
           'Sparse PW (t=0.7)']

selected_idx = [1, 12, 2, 4, 6, 8, 11]
selected_scores = np.array([scores[i] for i in selected_idx])
values = selected_scores
selected_stds = np.array([scores_std[i] for i in selected_idx])
selected_methods = [methods[i] for i in selected_idx]

plt.barh(selected_methods[::-1], values[::-1], 
         xerr=selected_stds, capsize=10, color=colormaps['tab20b'](np.arange(len(values))))
plt.axvline(x=0, color='red', linestyle='--', linewidth=2)

ax = plt.gca()  # Get current axis
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['bottom'].set_visible(True)

plt.xlabel('Average Test R-squared', fontsize=16)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.title('Ablation in Fine-tune Strategies', fontsize=16)
plt.xlim(0.55, 0.65)
plt.savefig('./figure/finetune_ablation.png', dpi=300, bbox_inches='tight')
plt.show()
```
