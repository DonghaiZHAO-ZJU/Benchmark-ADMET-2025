```python
import os
import pickle
import pandas as pd
import numpy as np
from collections import defaultdict
from sklearn.metrics import r2_score, mean_squared_error
from tqdm import tqdm
import matplotlib.pyplot as plt
from rdkit import Chem
from rdkit.Chem import AllChem
from rdkit import RDLogger
RDLogger.DisableLog('rdApp.*')

from MoleculeACE import MLP, GCN, SVM, MPNN, Data, Descriptors, calc_rmse, calc_cliff_rmse, get_benchmark_config
from molmcl.splitters import moleculeace_split, random_split
from rogi import RoughnessIndex

# Define the featurizer
get_fp = lambda smi: AllChem.GetMorganFingerprintAsBitVect(
    Chem.MolFromSmiles(smi), radius=2, nBits=1024)
```

```python
def rrmse(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    value = np.sqrt((np.sum((y_true - y_pred) ** 2) / np.sum(y_true ** 2)) / len(y_true))
    return value
```

```python
splits = 'stratified'  # [random, stratified]
assert splits in ['random', 'stratified']
```

```python
# Define inputs and models
descriptor = Descriptors.ECFP
algorithm = MLP
hyperparameters = {
    'dropout': 0,
    'epochs': 100,
    'hidden': 1024,
    'lr': 0.0005,
    'n_layers': 2
}

# descriptor = Descriptors.GRAPH
# algorithm = GCN
# hyperparameters = {
#     'dropout': 0,
#     'epochs': 100,
#     'fc_hidden': 256,
#     'lr': 5e-05,
#     'n_conv_layers': 5,
#     'n_fc_layers': 2,
#     'node_hidden': 512,
#     'transformer_hidden': 256
# }
```

```python
# Iterate through all datasets:
dataset2test_r2 = defaultdict(list)
for file in tqdm(os.listdir('./MoleculeACE/Data/benchmark_data')):
    if file.endswith('.csv'):
        dataset = file.split('.')[0]
        tqdm.write(dataset)

        # load dataset
        df = pd.read_csv(f'./data/finetune/moleculeace/{file}')

        # prepare train/val/test splits
        if splits == 'random':
            train_idx, val_idx, test_idx = random_split(df['smiles'])
        else:
            train_idx, val_idx, test_idx = moleculeace_split(df['smiles'], df['y'], val_size=0.1, test_size=0.1)

        x_train = np.array([get_fp(df['smiles'][i]) for i in (train_idx)], dtype=np.float32)
        y_train = np.array([df['y'][i] for i in (train_idx)], dtype=np.float32)
        x_val = np.array([get_fp(df['smiles'][i]) for i in (val_idx)], dtype=np.float32)
        y_val = np.array([df['y'][i] for i in (val_idx)], dtype=np.float32)
        x_test = np.array([get_fp(df['smiles'][i]) for i in (test_idx)], dtype=np.float32)
        y_test = np.array([df['y'][i] for i in (test_idx)], dtype=np.float32)

        # train 3 runs and take the average 
        test_scores = []
        for _ in range(5):
            # train models
            model = algorithm(**hyperparameters)
            model.train(x_train, y_train, x_val, y_val)
    
            # load best validation model
            with open('best_model.pkl', 'rb') as f:
                best_model = pickle.load(f)
            model.model = best_model
            
            # compute test r-square
            y_hat = model.predict(x_test)
            test_scores.append(r2_score(y_test, y_hat))
            
        dataset2test_r2[dataset] = np.mean(test_scores)

        tqdm.write(f'rogi = {dataset2rogi[dataset]}')
        tqdm.write(f'test_r2 = {dataset2test_r2[dataset]}')
        tqdm.write('-----')
```

```python
with open(f'MLP_{splits}.pk', 'wb') as f:
    pickle.dump(dataset2test_r2, f)
```

### Relationship between QSPR and learned prompt weight

```python
from sklearn.decomposition import PCA
```

```python
data = pd.read_csv('qspr_vs_pw.csv')
datasets = data['Dataset'].to_list()
pw = np.array([eval(s) for s in data['PW (optimized)']])
qspr = np.array([eval(s.replace(' ', ', ')) for s in data['QSPR']])
print(len(pw), len(qspr))
```

```python
sorted_idx = np.argsort(abs(qspr-1/3).sum(-1))[::-1]
pw = pw[sorted_idx]
qspr = qspr[sorted_idx]
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Number of vectors and vector length
num_vectors = pw.shape[0]
vector_length = pw.shape[1]

# Create a figure with 6x5 subplots
fig, axs = plt.subplots(6, 5, figsize=(20, 15))

# Flatten the axes array for easy iteration
axs = axs.flatten()

# Bar width
bar_width = 0.35

# Create side-by-side bar plots for each vector
for i in range(num_vectors):
    # X positions for the bars
    indices = np.arange(vector_length)
    
    # Plot bars for set1
    axs[i].bar(indices, pw[i], bar_width, label='Optimized PW', color='blue')
    
    # Plot bars for set2 with an offset to be side by side
    axs[i].bar(indices + bar_width, qspr[i], bar_width, label='QSPR', color='red')
    
    # Labeling the plot
    axs[i].set_xlabel('Positional Index')
    axs[i].set_ylabel('Value')
    axs[i].set_title(f'{datasets[i]}')
    axs[i].set_xticks(indices + bar_width / 2)
    axs[i].set_xticklabels([f'Pos {j+1}' for j in range(vector_length)])
    

# Creating a global legend
handles, labels = axs[0].get_legend_handles_labels()  # Get handles and labels from one of the plots
legend = fig.legend(handles, labels, loc='upper center', fontsize=16, ncol=2, frameon=True, facecolor='white')

# Adjust layout to prevent overlap
plt.tight_layout(rect=[0, 0, 1, 0.97])

# Show the combined plot with subplots
plt.show()
```
