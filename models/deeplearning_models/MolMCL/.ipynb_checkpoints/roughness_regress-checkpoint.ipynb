```python
def eval_func(model, val_loader, args, metric='r2'):
    assert metric in ['rmse', 'r2']
    model.eval()
    y_true, y_scores = [], []
    for step, batch in enumerate(val_loader):
        batch = batch.to(args.device)
        with torch.no_grad():
            predict = model(batch)

        y_true.append(batch.label.view(predict.shape))
        y_scores.append(predict)

    y_true = torch.cat(y_true, dim=0).cpu().numpy()
    y_scores = torch.cat(y_scores, dim=0).cpu().numpy()

    if 'CHEMBL' in data_name:
        if metric == 'rmse':
            score = -mean_squared_error(y_true, y_scores, squared=False)
        else:
            score = r2_score(y_true, y_scores)
    else:
        roc_list = []
        for i in range(y_true.shape[1]):
            # AUC is only defined when there is at least one positive data.
            if np.sum(y_true[:, i] == 1) > 0 and np.sum(y_true[:, i] == -1) > 0:
                is_valid = y_true[:, i] ** 2 > 0
                roc_list.append(roc_auc_score((y_true[is_valid, i] + 1) / 2, y_scores[is_valid, i]))

        # if len(roc_list) < y_true.shape[1]:
        #     print("Some target is missing!")
        #     print("Missing ratio: %f" %(1 - float(len(roc_list))/y_true.shape[1]))

        score = np.mean(roc_list)

    return score


def train_func(model, train_loader, criterion, optimizer, args):
    model.train()
    loss_history = []
    for idx, batch in enumerate(train_loader):
        batch.to(args.device)
        predict = model(batch)
        label = batch.label.view(predict.shape)

        if isinstance(criterion, nn.BCEWithLogitsLoss):
            mask = label == 0  # nan entry
            loss = criterion(predict.double(), (label + 1) / 2) * (~mask)
            loss = loss.sum() / (~mask).sum()
        elif isinstance(criterion, nn.MSELoss):
            loss = criterion(predict, label)
            loss = loss.mean()
        else:
            raise Exception

        optimizer.zero_grad()
        loss.backward()
        if args.gradient_clip > 0:
            nn.utils.clip_grad_norm_(model.parameters(), args.gradient_clip)
        optimizer.step()

        loss_history.append(loss.item())
        # print('[Epoch {}] Loss: {}'.format(epoch, round(np.mean(loss_history), 3)))
    return np.mean(loss_history)


def get_optimizer(model, pretrain_lr=0.001, finetune_lr=0.001, decay=1e-6):
    layer_list = []
    for name, param in model.named_parameters():
        if 'gnn' in name or 'aggrs' in name:
            layer_list.append(name)
    pretrain_params = list(
        map(lambda x: x[1], list(filter(lambda kv: kv[0] in layer_list, model.named_parameters()))))
    finetune_params = list(
        map(lambda x: x[1], list(filter(lambda kv: kv[0] not in layer_list, model.named_parameters()))))
    optimizer = torch.optim.Adam(
        [{'params': pretrain_params, 'lr': pretrain_lr}, {'params': finetune_params}],
        finetune_lr, weight_decay=decay
    )
    return optimizer

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
```

```python
def get_model_list(checkpoints, args, verbose=True):
    model_list = []
    for cpk_type, cpk_path in checkpoints:
        print('Method:', cpk_type)
        if cpk_type == 'grover':
            model_list.append(None) # build later
        else:
            if cpk_type == 'ours_rich':
                atom_feat_dim, bond_feat_dim = 170, 14
            else:
                atom_feat_dim, bond_feat_dim = None, None
            model = GNNPredictor(args.num_layer,
                                 args.emb_dim,
                                 1,
                                 JK=args.JK,
                                 drop_ratio=0,
                                 gnn_type=args.gnn_type,
                                 atom_feat_dim=atom_feat_dim,
                                 bond_feat_dim=bond_feat_dim,
                                 temperature=args.temperature,
                                 baseline=cpk_type,
                                 add_mean_pool=False,
                                 use_prompt='ours' in cpk_type)

            # Load pretrained checkpoint
            if cpk_type == 'ours' and cpk_path:
                if verbose:
                    print('  Loading checkpoint from {}'.format(cpk_path))
                model.load_state_dict(torch.load(cpk_path)['wrapper'], strict=False)

            elif cpk_path:
                if verbose:
                    print('  Loading gnn checkpoint from {}'.format(cpk_path))
                model.gnn.load_state_dict(torch.load(cpk_path, map_location='cpu'), strict=False)

            model.eval()
            model.to(args.device)
            model.freeze_aggr_module()
            model_list.append(model)
    return model_list
```

```python
def get_train_loader(data_name, args, few_shot_num=10, few_shot_ratio=0, few_shot_cliff=False, seed=42):
    dataset = MoleculeDataset(args.data_dir, data_name, basic_feat=args.basic_feat)
    num_task = dataset.num_task
    print('Loading dataset of size {} with {} target.'.format(len(dataset), num_task))
    
    # MoleculeACE stratified random split
    train_idx, test_idx, val_idx = moleculeace_split(dataset.smiles, dataset.labels, val_size=0.1, 
                                                     test_size=0.1, seed=seed)

    train_dataset, val_dataset, test_dataset = \
        Subset(dataset, train_idx), Subset(dataset, val_idx), Subset(dataset, test_idx)
    
    # train:
    train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
    train_full_loader = DataLoader(train_dataset, batch_size=len(train_dataset), shuffle=False)
    
    # val:
    val_loader = DataLoader(val_dataset, batch_size=len(val_dataset), shuffle=False)
    # tst:
    test_loader = DataLoader(test_dataset, batch_size=len(test_dataset), shuffle=False)
    # train-val:
    train_val_idx = train_idx + val_idx
    train_val_dataset = Subset(dataset, train_val_idx)
    train_val_loader = DataLoader(train_val_dataset, batch_size=len(train_val_dataset), shuffle=False)
    # full:
    full_idx = train_idx + val_idx + test_idx
    full_dataset = Subset(dataset, full_idx)
    full_loader = DataLoader(full_dataset, batch_size=len(full_dataset), shuffle=False)
    
    # extract train-cliff pairs:
    train_smiles = [dataset.smiles[i] for i in train_idx]
    train_labels = [dataset.labels[i] for i in train_idx]
    sim = moleculeace_similarity(train_smiles)
    fc = (get_fc(train_labels, in_log10=True) > 10).astype(int)
    cliffs = np.logical_and(sim == 1, fc == 1).astype(int)
    cliff_pair_inds = np.argwhere(cliffs)
    noncliffs = np.logical_and(sim == 1, fc == 0).astype(int)
    noncliff_pair_inds = np.argwhere(noncliffs)    
    
    # few_shot:
    if few_shot_num > 0 or few_shot_ratio > 0:
        if few_shot_ratio > 0:
            few_shot_num = int(len(train_idx) * few_shot_ratio)
            
        train_fps = np.array([np.array(get_fp(smi)) for smi in train_smiles])

        if few_shot_num < len(train_idx):
            kmeans = KMeans(n_clusters=few_shot_num).fit(train_fps)
            few_shot_idx = np.argmin(euclidean_distances(train_fps, kmeans.cluster_centers_), axis=0)
            few_shot_idx = set(few_shot_idx)
            few_shot_idx = [train_idx[i] for i in few_shot_idx]
        else:
            few_shot_idx = train_idx

        few_shot_dataset = Subset(dataset, few_shot_idx)
        few_shot_loader = DataLoader(few_shot_dataset, batch_size=len(few_shot_dataset), shuffle=True)
    else:
        few_shot_loader = None

    return dataset, (train_idx, val_idx, test_idx), (noncliff_pair_inds, cliff_pair_inds), (train_loader, \
            val_loader, test_loader), full_loader, train_val_loader, train_full_loader, few_shot_loader
```

```python
get_fp = lambda key: AllChem.GetMorganFingerprintAsBitVect(
    Chem.MolFromSmiles(key), radius=2, nBits=512)

def sample_triplet(cliff_pair_inds, noncliff_pair_inds, seed=-1):
    if seed > -1:
        np.random.seed(seed)
    while True:
        c_idx = np.random.choice(np.arange(len(cliff_pair_inds)))
        c_pair = cliff_pair_inds[c_idx]
        nc_idx = np.argwhere(c_pair[0] == noncliff_pair_inds[:, 0])
        if len(nc_idx):
            nc_idx = np.random.choice(nc_idx.flatten())
            nc_pair = noncliff_pair_inds[nc_idx]
            break
    return c_pair, nc_pair
```

```python
def get_reps_local(ori_model, loader, method, best_checkpoint=None):
    model = copy.deepcopy(ori_model)
    if best_checkpoint is not None:
        model.load_state_dict(best_checkpoint)
    model.eval()
    for batch in loader:
        batch.to(args.device)
        with torch.no_grad():
            h_g, node_repres = model.gnn(batch.x, batch.edge_index, batch.edge_attr, batch.batch)
            if method == 'ours':            
                graph_reps = []
                for j in range(len(model.prompt_token)):
                    h_g, h_x, _ = model.aggrs[j](node_repres, batch.batch)
                    graph_reps.append(h_g)
                graph_reps = torch.stack(graph_reps)
                prompt_weight = F.softmax(model.prompt_weight / model.temperature, dim=-1)
                graph_rep = torch.matmul(graph_reps.transpose(0, 2), prompt_weight).transpose(0, 1)
            elif method == 'ours_sig':
                graph_rep, _, _ = model.aggrs[0](node_repres, batch.batch)
            else:
                graph_rep = h_g
    smi2rep = {batch.smi[k]: graph_rep[k].cpu() for k in range(graph_rep.shape[0])}
    smi2label = {batch.smi[k]: batch.label[k].item() for k in range(graph_rep.shape[0])}
    return smi2rep, smi2label

checkpoints = [
    # ('ours', './checkpoint/pretrain/zinc_mcl_basic.pt'),
    # ('graphlog', '../../GraphLoG/models/graphlog.pth'),
    ('molclr', './checkpoint/pretrain/molclr_gin.pth'),
]
args.prompt_optim = 'ri'
args.temperature = 1
ri_metric = 'euclidean' # euclidean, cosine
data_name = 'CHEMBL237_Ki'  # CHEMBL237_Ki
initial_prompt_weight = None

# get dataloader:
res = get_train_loader(data_name, args, few_shot_num=0, few_shot_ratio=0, few_shot_cliff=False)
dataset = res[0]
train_idx, val_idx, test_idx = res[1]
noncliff_pair_inds, cliff_pair_inds = res[2]
train_loader, val_loader, test_loader = res[3]
full_loader, train_val_loader, train_full_loader, few_shot_loader = res[4], res[5], res[6], res[7]
print('Number of cliff pairs:', len(cliff_pair_inds))
```

```python
# get model list:
model_list = get_model_list(checkpoints, args)

# retrieve representation:
graph_rep_dict, graph_label_dict = defaultdict(dict), defaultdict(dict)
r2_history_dict = defaultdict(list)
for i in tqdm(range(len(model_list))):
    print('Build representation from {}.'.format(checkpoints[i][0]))
    model = model_list[i]
    method = checkpoints[i][0]
    
    if method == 'ours':
        # roughness index optimized prompt weight
        set_seed(42)
        model.eval()
        for batch in train_val_loader:
            batch.to(args.device)
            with torch.no_grad():
                h_g, node_repres = model.gnn(batch.x, batch.edge_index, batch.edge_attr, batch.batch)
                graph_reps = []
                for j in range(len(model.prompt_token)):
                    h_g, h_x, _ = model.aggrs[j](node_repres, batch.batch)
                    graph_reps.append(h_g)
                graph_reps = torch.stack(graph_reps)
            labels = batch.label.view(-1, model.num_tasks)
        prompt_weight = optimize_prompt_weight_ri(graph_reps.cpu(), labels.cpu(), n_runs=50, 
                                                  n_inits=50, n_points=5, n_restarts=512,
                                                  n_samples=512, temperature=model.temperature, 
                                                  metric=ri_metric, skip_bo=False, verbose=False)
        model.set_prompt_weight(prompt_weight.to(args.device))
        ipw = copy.deepcopy(model.get_prompt_weight(softmax=True).detach().cpu().numpy())
        print('Initial prompt weight:', ipw)
        
    set_seed(42)
    criterion = nn.MSELoss(reduction='none')
    optimizer = get_optimizer(model, pretrain_lr=0.0005, finetune_lr=0.0005)
    best_score, best_checkpoint = -float('inf'), None
    for ep in tqdm(range(101)):
        if ep in [0, 10, 20, 50, 80, 100]:
            smi2rep, smi2label = get_reps_local(model, full_loader, method, best_checkpoint)
            graph_rep_dict['{}-{}'.format(method, ep)] = smi2rep
            graph_label_dict['{}-{}'.format(method, ep)] = smi2label
            
            X = torch.stack(list(smi2rep.values())).numpy()[:len(train_idx)]
            Y = np.array(list(smi2label.values()))[:len(train_idx)]
            rogi = RoughnessIndex(Y=Y, X=X, metric=ri_metric, verbose=False)
            ri = rogi.compute_index()
            print('Roughness Index (ep{}): {}'.format(ep, ri))
            
            if ep == 100:
                break
                    
        loss = train_func(model, train_loader, criterion, optimizer, args)
        train_score = eval_func(model, train_loader, args)
        val_score = eval_func(model, val_loader, args)
        test_score = eval_func(model, test_loader, args)
        r2_history_dict[method].append((train_score, val_score, test_score))
        
        if -loss > best_score:
            best_score = -loss
            best_checkpoint = copy.deepcopy(model.state_dict())
        
        if ep % 10 == 0:
            print('[epoch {}] Train R2: {} Val R2: {}'.format(ep, train_score, val_score))
            
    with open('./inter/new-{}-{}.pk'.format(data_name, method), 'wb') as f:
        pickle.dump([graph_rep_dict, graph_label_dict, r2_history_dict], f)
        
print('complete.')
# 1796
```

```python
# compute roughness index
perplexity = 25
n_clusters = 10
cliff_distances_dict = defaultdict(list)
# tsne_seeds = [[[42, 43, 44, 44, 44], [42, 43, 48, 48, 46]],
#               [[42, 44, 48, 44, 44], [42, 44, 47, 44, 44]],
#               [[42, 43, 48, 48, 50], [42, 43, 49, 49, 51]]]
tsne_seeds = [[[42, 44, 48, 44, 44], [42, 44, 47, 44, 44]]]

# for mi, method in enumerate(['molclr', 'graphlog', 'ours']):

cliff_dist_list, noncliff_dist_list = [], []

for mi, method in enumerate(['ours']):
    
    with open('./Robustac/inter/{}-{}.pk'.format(data_name, method), 'rb') as f:
        graph_rep_dict, graph_label_dict, r2_history_dict = pickle.load(f)
    
    fig, axes = plt.subplots(2, 5, figsize=(25, 6))
    ((axt1, axt2, axt3, axt4, axt5), (axv1, axv2, axv3, axv4, axv5)) = axes
    for ei, ep in enumerate([0, 10, 20, 50, 100]):
        ei += 1
        key = '{}-{}'.format(method, ep)
        print('Key:', key)

        X = torch.stack(list(graph_rep_dict[key].values())).numpy()
        Y = np.array(list(graph_label_dict[key].values()))
        X_train, X_val = X[:len(train_idx)], X[len(train_idx):len(train_idx)+len(val_idx)]
        Y_train, Y_val = Y[:len(train_idx)], Y[len(train_idx):len(train_idx)+len(val_idx)]

        # 2d-plot:
        ## Subplot-1:
        
        rogi = RoughnessIndex(Y=Y_train, X=X_train, metric=ri_metric, verbose=False)
        ri = rogi.compute_index()
        print('Roughness Index:', ri)

        ### compute fingerprint clustering:
        X_smi = list(graph_rep_dict[key].keys())[:len(train_idx)]
        X_fps = np.array([np.array(get_fp(smi)) for smi in X_smi])
        kmeans = KMeans(n_clusters=n_clusters, random_state=42).fit(X_fps)
        kmeans_label = kmeans.labels_
        kmeans = KMeans(n_clusters=n_clusters, random_state=42).fit(X_train)
        kmeans_label2 = kmeans.labels_

        ### dimension reduction:
        tsne = TSNE(n_components=2, learning_rate='auto', init='random',
                    metric='precomputed', random_state=tsne_seeds[mi][0][ei-1], perplexity=perplexity)
        X_2d = tsne.fit_transform(squareform(rogi._Dx))

        ### main plot:        
        scatter1 = eval('axt{}'.format(ei)).scatter(
            X_2d[:, 0], X_2d[:, 1], c=rogi._Y, s=40, cmap=cm.get_cmap("coolwarm"), 
            alpha=0.8, zorder=10, edgecolors='grey', linewidths=0.4)
            
        for ci in set(kmeans_label):
            points = X_2d[kmeans_label == ci]
            hull = ConvexHull(points)
            x_hull = np.append(points[hull.vertices,0],
                               points[hull.vertices,0][0])
            y_hull = np.append(points[hull.vertices,1],
                               points[hull.vertices,1][0])
            # interpolate
            dist = np.sqrt((x_hull[:-1] - x_hull[1:])**2 + (y_hull[:-1] - y_hull[1:])**2)
            dist_along = np.concatenate(([0], dist.cumsum()))
            spline, u = interpolate.splprep([x_hull, y_hull], u=dist_along, s=0, per=1)
            interp_d = np.linspace(dist_along[0], dist_along[-1], 50)
            interp_x, interp_y = interpolate.splev(interp_d, spline)
            # plot shape
            eval('axt{}'.format(ei)).fill(interp_x, interp_y, '--', c='lightgrey', 
                     edgecolor='grey', linewidth=1, alpha=0.3, zorder=9)
            
        plot_text = 'ROGI: {}\nRand Index: {}'.format(
            round(ri, 3),
            round(adjusted_rand_score(kmeans_label2, kmeans_label), 3))  
        print(plot_text)
        eval('axt{}'.format(ei)).text(0.03, 0.05, plot_text, fontsize=12,
            transform=eval('axt{}'.format(ei)).transAxes, zorder=20,
            bbox=dict(facecolor='white', edgecolor='grey', boxstyle='round', alpha=0.7))
        eval('axt{}'.format(ei)).tick_params(
            left=False, right=False, labelleft=False, labelbottom=False, bottom=False)

        ## Subplot-2:
        rogi = RoughnessIndex(Y=Y_val, X=X_val, metric=ri_metric, verbose=False)
        ri = rogi.compute_index()
        
        ### dimension reduction:
        tsne = TSNE(n_components=2, learning_rate='auto', init='random',
                    metric='precomputed', random_state=tsne_seeds[mi][1][ei-1], perplexity=perplexity)
        X_2d = tsne.fit_transform(squareform(rogi._Dx))

        # find test cliffs:
        val_smiles = [dataset.smiles[i] for i in val_idx]
        val_labels = [dataset.labels[i] for i in val_idx]
        # val_smiles = list(graph_rep_dict[key].keys())[len(train_idx):len(train_idx)+len(val_idx)]
        # val_labels = Y[len(train_idx):len(train_idx)+len(val_idx)]
  
        sim = moleculeace_similarity(val_smiles)
        fc = (get_fc(val_labels, in_log10=True) > 10).astype(int)
        cliffs = np.logical_and(sim == 1, fc == 1).astype(int)
        cliff_pair_inds = np.unique(np.argwhere(cliffs), axis=0)
        cliff_distances = []
        for idx_1, idx_2 in cliff_pair_inds:
            cliff_distances.append(squareform(rogi._Dx)[idx_1, idx_2])
        noncliffs = np.logical_and(sim == 1, fc == 0).astype(int)
        noncliff_pair_inds = np.unique(np.argwhere(noncliffs), axis=0)
        noncliff_distances = []
        for idx_1, idx_2 in noncliff_pair_inds:
            noncliff_distances.append(squareform(rogi._Dx)[idx_1, idx_2])

        cliff_dist_list.append(cliff_distances)
        noncliff_dist_list.append(noncliff_distances)
        
        print('Average cliff pair distance:', np.mean(cliff_distances))
        print('Average noncliff pair distance:', np.mean(noncliff_distances))
        print('Cliff-noncliff distance ratio:', np.mean(cliff_distances) / np.mean(noncliff_distances))
        print('----------')
        c_pair, nc_pair = sample_triplet(cliff_pair_inds, noncliff_pair_inds, seed=12) # 12
        
        # n_clusters = len(set(kmeans_label))
        # cmap = cm.get_cmap("rainbow_r", n_clusters)
        cmap = cm.get_cmap("coolwarm")
        scatter2 = eval('axv{}'.format(ei)).scatter(X_2d[:, 0], X_2d[:, 1], c=rogi._Y, s=40, 
                              cmap=cmap, alpha=0.8, zorder=10, edgecolors='grey', linewidths=0.4)
        
        eval('axv{}'.format(ei)).arrow(X_2d[c_pair[0], 0], X_2d[c_pair[0], 1], 
                  X_2d[c_pair[1], 0]-X_2d[c_pair[0], 0], X_2d[c_pair[1], 1]-X_2d[c_pair[0], 1],
                  color='red', alpha=0.7, zorder=12, width=0.3, head_width=1.3)
        eval('axv{}'.format(ei)).arrow(X_2d[nc_pair[0], 0], X_2d[nc_pair[0], 1], 
                  X_2d[nc_pair[1], 0]-X_2d[nc_pair[0], 0], X_2d[nc_pair[1], 1]-X_2d[nc_pair[0], 1],
                  color='blue', alpha=0.7, zorder=12, width=0.3, head_width=1.2)
        if ep == 0:
            r2 = 'N/A'
        else:
            r2 = round(r2_history_dict[method][ep-1][1], 3)  
        dist_ratio = np.mean(cliff_distances) / np.mean(noncliff_distances)
        plot_text = 'R-Squared: {}\nCliff-noncliff Distance Ratio: {}'.format(r2, round(dist_ratio, 3))
        eval('axv{}'.format(ei)).text(0.03, 0.05, plot_text, fontsize=12,
            transform=eval('axv{}'.format(ei)).transAxes, zorder=12,
            bbox=dict(facecolor='white', edgecolor='grey', boxstyle='round', alpha=0.7))

        eval('axv{}'.format(ei)).tick_params(
            left=False, right=False, labelleft=False, labelbottom=False, bottom=False)
        
    plt.tight_layout()
    plt.colorbar(scatter2, ax=axes.ravel().tolist())        
    
    # plt.savefig('./figure/{}-{}-fewepoch.png'.format(data_name, key), dpi=300)
    plt.show()
    break
```

```python
from robustac.finetune.model import GNNPredictor
def get_reps_local(ori_model, loader, method, best_checkpoint=None):
    smi_reps, smi_labels = [], []
    model = copy.deepcopy(ori_model)
    if best_checkpoint is not None:
        model.load_state_dict(best_checkpoint)
    model.eval()
    for batch in loader:
        batch.to(args.device)
        with torch.no_grad():
            h_g, node_repres = model.gnn(batch.x, batch.edge_index, batch.edge_attr, batch.batch)
            if method == 'ours':            
                graph_reps = []
                for j in range(len(model.prompt_token)):
                    h_g, h_x, _ = model.aggrs[j](node_repres, batch.batch)
                    graph_reps.append(h_g)
                graph_reps = torch.stack(graph_reps)
                
                for gi in range(graph_reps.shape[0]):
                    graph_rep_sig = graph_reps[gi]
                    smi_reps.append({batch.smi[k]: graph_rep_sig[k].cpu() for k in range(graph_rep_sig.shape[0])})                
                
                prompt_weight = F.softmax(model.prompt_weight / model.temperature, dim=-1)
                graph_rep = torch.matmul(graph_reps.transpose(0, 2), prompt_weight).transpose(0, 1)
            else:
                graph_rep = h_g
                
    smi_reps.append({batch.smi[k]: graph_rep[k].cpu() for k in range(graph_rep.shape[0])})
    smi_labels = [{batch.smi[k]: batch.label[k].item() for k in range(graph_rep.shape[0])}]
    return smi_reps, smi_labels
```

```python
checkpoints = [
    ('ours', './checkpoint/pretrain/zinc_mcl_basic.pt'),
    ('molclr', '../../MolCLR/ckpt/pretrained_gin/checkpoints/model.pth'),
]
args.prompt_optim = 'ri'
args.temperature = 1
ri_metric = 'euclidean' 
args.data_dir = './data/finetune/moleculeace'
data_name = 'CHEMBL237_Ki'  # CHEMBL237_Ki CHEMBL262_Ki
initial_prompt_weight = None

# get dataloader:
n_clusters = 5
res = get_train_loader(data_name, args, few_shot_num=0, few_shot_ratio=0, few_shot_cliff=False, seed=42)
dataset = res[0]
train_idx, val_idx, test_idx = res[1]
noncliff_pair_inds, cliff_pair_inds = res[2]
train_loader, val_loader, test_loader = res[3]
full_loader, train_val_loader, train_full_loader, few_shot_loader = res[4], res[5], res[6], res[7]
print('Number of cliff pairs:', len(cliff_pair_inds))
    
# get model list:
model_list = get_model_list(checkpoints, args)

# retrieve representation:
cluster_label_start = []
ep2randidx = defaultdict(list)
ep2val_auc = defaultdict(list)
for i in tqdm(range(len(model_list))):
    print('Build representation from {}.'.format(checkpoints[i][0]))
    model = model_list[i]
    method = checkpoints[i][0]
    
    if method == 'ours':
        # roughness index optimized prompt weight
        model.eval()
        for batch in train_val_loader:
            batch.to(args.device)
            with torch.no_grad():
                h_g, node_repres = model.gnn(batch.x, batch.edge_index, batch.edge_attr, batch.batch)
                graph_reps = []
                for j in range(len(model.prompt_token)):
                    h_g, h_x, _ = model.aggrs[j](node_repres, batch.batch)
                    graph_reps.append(h_g)
                graph_reps = torch.stack(graph_reps)
            labels = batch.label.view(-1, model.num_tasks)
        
        prompt_weight = optimize_prompt_weight_ri(graph_reps.cpu(), labels.cpu(), n_runs=50, 
                                                  n_inits=500, n_points=5, n_restarts=512,
                                                  n_samples=512, temperature=model.temperature, 
                                                  metric=ri_metric, skip_bo=True, verbose=False)
        model.set_prompt_weight(prompt_weight.to(args.device))
        print('Initial prompt weight       :', prompt_weight)
        ipw = copy.deepcopy(model.get_prompt_weight(softmax=True).detach().cpu().numpy())
        model.prompt_weight.requires_grad = False
        print('Initial prompt weight (prob):', ipw)
        
        
    set_seed(42)
    if 'CHEMBL' in data_name:
        criterion = nn.MSELoss(reduction='none')
        optimizer = get_optimizer(model, pretrain_lr=0.0005, finetune_lr=0.0005)
    else:
        criterion = nn.BCEWithLogitsLoss(reduction="none")
        optimizer = get_optimizer(model, pretrain_lr=0.0001, finetune_lr=0.001)

    
    best_score, best_checkpoint = -float('inf'), None
    cluster_label_start = []
    for ep in range(101):
        # if ep % 2 == 0:
        smi_reps, smi_labels = get_reps_local(model, train_full_loader, method, best_checkpoint)

        cluster_label_ep = []
        for smi2rep in smi_reps:
            X_train = torch.stack(list(smi2rep.values())).numpy()
            kmeans = KMeans(n_clusters=n_clusters, random_state=42).fit(X_train)
            if ep == 0:
                cluster_label_start.append(kmeans.labels_)
            cluster_label_ep.append(kmeans.labels_)

        for i in range(len(cluster_label_start)):
            rand_index = adjusted_rand_score(cluster_label_start[i], cluster_label_ep[i])
            print('[ep{}] {}'.format(ep, rand_index))

            ep2randidx[ep].append(rand_index)
                    
        loss = train_func(model, train_loader, criterion, optimizer, args)
        val_score = eval_func(model, val_loader, args)
        ep2val_auc[ep].append(val_score)
        
        if -loss > best_score:
            best_score = -loss
            best_checkpoint = copy.deepcopy(model.state_dict())
        
        if ep % 10 == 0:
            print('[epoch {}] Train R2: {} Val R2: {}'.format(ep, train_score, val_score))
            
    if method == 'ours':
        fpw = copy.deepcopy(model.get_prompt_weight(softmax=True).detach().cpu().numpy())
```

## Visualization (0-shot --> all-shot):

```python
def get_reps_local(ori_model, loader, method, best_checkpoint=None):
    model = copy.deepcopy(ori_model)
    if best_checkpoint is not None:
        model.load_state_dict(best_checkpoint)
    model.eval()
    for batch in loader:
        batch.to(args.device)
        with torch.no_grad():
            h_g, node_repres = model.gnn(batch.x, batch.edge_index, batch.edge_attr, batch.batch)
            if method == 'ours':            
                graph_reps = []
                for j in range(len(model.prompt_token)):
                    h_g, h_x, _ = model.aggrs[j](node_repres, batch.batch)
                    graph_reps.append(h_g)
                graph_reps = torch.stack(graph_reps)
                prompt_weight = F.softmax(model.prompt_weight / model.temperature, dim=-1)
                graph_rep = torch.matmul(graph_reps.transpose(0, 2), prompt_weight).transpose(0, 1)
            elif method == 'ours_sig':
                graph_rep, _, _ = model.aggrs[0](node_repres, batch.batch)
            else:
                graph_rep = h_g
    smi2rep = {batch.smi[k]: graph_rep[k].cpu() for k in range(graph_rep.shape[0])}
    smi2label = {batch.smi[k]: batch.label[k].item() for k in range(graph_rep.shape[0])}
    return smi2rep, smi2label
```

```python
checkpoints = [
    ('ours', './checkpoint/pretrain/zinc_mcl_basic.pt'),
    ('graphlog', '../../GraphLoG/models/graphlog.pth'),
    ('molclr', '../../MolCLR/ckpt/pretrained_gin/checkpoints/model.pth'),
]
args.prompt_optim = 'ri'
args.temperature = 1
ri_metric = 'euclidean'
data_name = 'CHEMBL237_Ki'
initial_prompt_weight = None
# few_shot_ratio_grid = [0.01, 0.05, 0.1, 0.5, 1]
few_shot_ratio_grid = [0]

graph_rep_dict, graph_label_dict = {}, {}
r2_history_dict = defaultdict(list)
for i, few_shot_num in enumerate(few_shot_ratio_grid):
    few_shot_ratio = int(few_shot_ratio_grid[i] * 100)
    few_shot_num = int(few_shot_ratio_grid[i]*len(train_idx))
    few_shot_epoch = 50
    
    # get dataloader:
    res = get_train_loader(data_name, args, few_shot_num=few_shot_num, 
                           few_shot_ratio=0, few_shot_cliff=False, seed=1)
    dataset = res[0]
    train_idx, val_idx, test_idx = res[1]
    noncliff_pair_inds, cliff_pair_inds = res[2]
    train_loader, val_loader, test_loader = res[3]
    full_loader, train_val_loader, train_full_loader, few_shot_loader = res[4], res[5], res[6], res[7]
    
    # get model list:
    model_list = get_model_list(checkpoints, args)
    
    # few-shot training
    for mi, model in enumerate(model_list):
        method = checkpoints[mi][0]
        print('Few-shot({}) training with method {}:'.format(few_shot_num, method))
        if method == 'ours' and few_shot_num > 0:
            # roughness index optimized prompt weight
            model.eval()
            for batch in few_shot_loader:
                batch.to(args.device)
                with torch.no_grad():
                    h_g, node_repres = model.gnn(batch.x, batch.edge_index, batch.edge_attr, batch.batch)
                    graph_reps = []
                    for j in range(len(model.prompt_token)):
                        h_g, h_x, _ = model.aggrs[j](node_repres, batch.batch)
                        graph_reps.append(h_g)
                    graph_reps = torch.stack(graph_reps)
                labels = batch.label.view(-1, model.num_tasks)

            init_prompt_weight = optimize_prompt_weight_ri(graph_reps.cpu(), labels.cpu(), n_runs=50, 
                                                           n_inits=100, n_points=5, n_restarts=512,
                                                           n_samples=512, temperature=model.temperature, 
                                                           metric=ri_metric, skip_bo=True, verbose=False)
            
            model.set_prompt_weight(init_prompt_weight.to(args.device))
            print('Initial prompt weight       :', init_prompt_weight)
            ipw = copy.deepcopy(model.get_prompt_weight(softmax=True).detach().cpu().numpy())
            model.prompt_weight.requires_grad = False
            print('Initial prompt weight (prob):', ipw)
            
        
        if few_shot_num > 0:
            set_seed(42)
            model.train()
            criterion = nn.MSELoss(reduction='none')
            optimizer = get_optimizer(model, pretrain_lr=0.0005, finetune_lr=0.0005)
            best_score, best_checkpoint = -float('inf'), None
            for bi in range(few_shot_epoch):
                score = -train_func(model, few_shot_loader, criterion, optimizer, args)
                val_score = eval_func(model, val_loader, args)

                if score > best_score:
                    best_score = score
                    best_checkpoint = copy.deepcopy(model.state_dict())

                r2_history_dict['{}-{}'.format(method, few_shot_ratio)].append(val_score)
        
        # retrieve representation:
        smi2rep, smi2label = get_reps_local(model, train_val_loader, method, best_checkpoint)
        graph_rep_dict['{}-{}'.format(method, few_shot_ratio)] = smi2rep
        graph_label_dict['{}-{}'.format(method, few_shot_ratio)] = smi2label
```

```python
perplexity = 5
for target_method in ['ours', 'graphlog', 'molclr']:
    fig, axes = plt.subplots(1, len(few_shot_num_grid), figsize=(20, 3))
    ax0, ax1, ax2, ax3, ax4 = axes
    for key in graph_rep_dict:
        method = key.split('-')[0]
        few_shot_ratio = int(float(key.split('-')[-1]))
        if method != target_method:
            continue
            
        fsi = np.argwhere(few_shot_ratio/100 == np.array(few_shot_ratio_grid)).item()

        X_val = torch.stack(list(graph_rep_dict[key].values())).numpy()[len(train_idx):]
        Y_val = np.array(list(graph_label_dict[key].values()))[len(train_idx):]

        rogi = RoughnessIndex(Y=Y_val, X=X_val, metric=ri_metric, verbose=False)
        ri = rogi.compute_index()
        # print('Roughness Index:', ri)

        ### dimension reduction:
        tsne = TSNE(n_components=2, learning_rate='auto', init='random',
                    metric='precomputed', random_state=0, perplexity=perplexity)
        X_2d = tsne.fit_transform(squareform(rogi._Dx))

        ### compute fingerprint clustering:
        X_smi = list(graph_rep_dict[key].keys())[len(train_idx):]
        X_fps = np.array([np.array(get_fp(smi)) for smi in X_smi])
        kmeans = KMeans(n_clusters=n_clusters, random_state=42).fit(X_fps)
        kmeans_label = kmeans.labels_
        kmeans = KMeans(n_clusters=n_clusters, random_state=42).fit(X_val)
        kmeans_label2 = kmeans.labels_
        
        ### compute cliff-noncliff distance
        val_smiles = [dataset.smiles[i] for i in val_idx]
        val_labels = [dataset.labels[i] for i in val_idx]
        sim = moleculeace_similarity(val_smiles)
        fc = (get_fc(val_labels, in_log10=True) > 10).astype(int)
        cliffs = np.logical_and(sim == 1, fc == 1).astype(int)
        cliff_pair_inds = np.unique(np.argwhere(cliffs), axis=0)
        cliff_distances = []
        for idx_1, idx_2 in cliff_pair_inds:
            cliff_distances.append(squareform(rogi._Dx)[idx_1, idx_2])
        noncliffs = np.logical_and(sim == 1, fc == 0).astype(int)
        noncliff_pair_inds = np.unique(np.argwhere(noncliffs), axis=0)
        noncliff_distances = []
        for idx_1, idx_2 in noncliff_pair_inds:
            noncliff_distances.append(squareform(rogi._Dx)[idx_1, idx_2])
        # print('Average cliff pair distance:', np.mean(cliff_distances))
        # print('Average noncliff pair distance:', np.mean(noncliff_distances))
        # print('Cliff-noncliff distance ratio:', np.mean(cliff_distances) / np.mean(noncliff_distances))
        # print('----------')

        scatter1 = eval('ax{}'.format(fsi)).scatter(
                    X_2d[:, 0], X_2d[:, 1], c=rogi._Y, s=40, cmap=cm.get_cmap("coolwarm"), 
                    alpha=0.8, zorder=10, edgecolors='grey', linewidths=0.4)

        plot_text = 'Rand Index: {}\nCliff-noncliff Distance Ratio: {}'.format(
            round(adjusted_rand_score(kmeans_label2, kmeans_label), 3),
            round(np.mean(cliff_distances) / np.mean(noncliff_distances), 3)
        )  
        eval('ax{}'.format(fsi)).text(0.03, 0.05, plot_text, fontsize=12,
            transform=eval('ax{}'.format(fsi)).transAxes, zorder=20,
            bbox=dict(facecolor='white', edgecolor='grey', boxstyle='round', alpha=0.7))
        eval('ax{}'.format(fsi)).tick_params(
            left=False, right=False, labelleft=False, labelbottom=False, bottom=False)
        
    plt.tight_layout()
    # plt.colorbar(scatter1, ax=axes.ravel().tolist())        
    # plt.savefig('./figure/fewshot-{}-{}.png'.format(data_name, target_method), dpi=300)
    plt.show()
```

## Convergence rate:

```python
checkpoints = [
    ('ours_sig_0', './checkpoint/pretrain/zinc_mcl_basic.pt'),
    ('ours_sig_1', './checkpoint/pretrain/zinc_mcl_basic.pt'),
    ('ours_sig_2', './checkpoint/pretrain/zinc_mcl_basic.pt'),
    ('ours', './checkpoint/pretrain/zinc_mcl_basic.pt'),
    ('graphlog', '../../GraphLoG/models/graphlog.pth'),
    ('molclr', '../../MolCLR/ckpt/pretrained_gin/checkpoints/model.pth'),
]
args.prompt_optim = 'ri'
args.temperature = 1
ri_metric = 'euclidean' 
args.data_dir = './data/finetune/moleculeace'
data_names = [
    'CHEMBL262_Ki',
    'CHEMBL237_Ki',
    'CHEMBL234_Ki',
    'CHEMBL2034_Ki'
]
data_stats = []
for data_name in data_names:
    print(data_name)
    # get dataloader:
    res = get_train_loader(data_name, args, few_shot_num=0, few_shot_ratio=0, few_shot_cliff=False, seed=42)
    dataset = res[0]
    train_idx, val_idx, test_idx = res[1]
    noncliff_pair_inds, cliff_pair_inds = res[2]
    train_loader, val_loader, test_loader = res[3]
    full_loader, train_val_loader, train_full_loader, few_shot_loader = res[4], res[5], res[6], res[7]
    print('Number of cliff pairs:', len(cliff_pair_inds))

    # get model list:
    model_list = get_model_list(checkpoints, args)
    
    r2_history_dict = defaultdict(list)
    for mi, model in tqdm(enumerate(model_list), total=len(model_list)):
        method = checkpoints[mi][0]
        if method == 'ours':
            # roughness index optimized prompt weight
            model.eval()
            for batch in train_val_loader:
                batch.to(args.device)
                with torch.no_grad():
                    h_g, node_repres = model.gnn(batch.x, batch.edge_index, batch.edge_attr, batch.batch)
                    graph_reps = []
                    for j in range(len(model.prompt_token)):
                        h_g, h_x, _ = model.aggrs[j](node_repres, batch.batch)
                        graph_reps.append(h_g)
                    graph_reps = torch.stack(graph_reps)
                labels = batch.label.view(-1, model.num_tasks)

            init_prompt_weight = optimize_prompt_weight_ri(graph_reps.cpu(), labels.cpu(), n_runs=50, 
                                                           n_inits=100, n_points=5, n_restarts=512,
                                                           n_samples=512, temperature=model.temperature, 
                                                           metric=ri_metric, skip_bo=True, verbose=False)

            model.set_prompt_weight(init_prompt_weight.to(args.device))
            print('Initial prompt weight       :', init_prompt_weight)
            ipw = copy.deepcopy(model.get_prompt_weight(softmax=True).detach().cpu().numpy())
            model.prompt_weight.requires_grad = False
            print('Initial prompt weight (prob):', ipw)

        set_seed(42)
        criterion = nn.MSELoss(reduction='none')
        optimizer = get_optimizer(model, pretrain_lr=0.0005, finetune_lr=0.0005)
        model.train()
        best_score, best_checkpoint = -float('inf'), None
        for bi in range(few_shot_epoch):
            train_func(model, train_loader, criterion, optimizer, args)
            train_score = eval_func(model, train_loader, args)
            val_score = eval_func(model, val_loader, args)
            r2_history_dict[method].append((train_score, val_score))
                
    data_stats.append(r2_history_dict)
```
