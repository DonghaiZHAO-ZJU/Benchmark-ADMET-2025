```python
# function for functional group featurization
fr_SMARTS = {}
fr_PATT = {}
with open(os.path.join(RDConfig.RDDataDir, 'FragmentDescriptors.csv'), 'r') as f:
    for line in f.readlines():
        if len(line) and line[0] != '#':
            splitL = line.rstrip().split('\t')
            if len(splitL) >= 3:
                name, descr, sma = splitL[:3]
                name = name.replace('=', '_')
                name = name.replace('-', '_')
                descr = descr.replace('"', '')
                patt = Chem.MolFromSmarts(sma)
                if patt is None or patt.GetNumAtoms() == 0:
                    print('Smarts %s could not be parsed' % (repr(sma)))
                    raise ImportError
                fr_SMARTS[name] = sma
                fr_PATT[name] = patt
fr_FUNCS = {name: func for name, func in Descriptors.descList if name in fr_SMARTS}

def generate_motif_features(mol, binary=True, verbose=False):
    features = np.zeros(len(fr_FUNCS), dtype=int)
    for i, fr in enumerate(fr_FUNCS):
        features[i] = fr_FUNCS[fr](mol)
        if binary:
            features[i] = int(features[i] > 0)
            if features[i] > 0 and verbose:
                print('Match {}: {}'.format(fr, fr_SMARTS[fr]))
            elif verbose:
                print('Unmatch {}'.format(fr))
    return features
    
def get_multi_repres(smi):
    mol = Chem.MolFromSmiles(smi)
    scaff_mol = get_scaffold(smi, return_mol=True)
    
    mol_motif = generate_motif_features(mol, binary=True)
    return mol_motif, get_fp(mol), get_fp(scaff_mol)

def tanimoto(v1, v2):
    if np.bitwise_and(v1, v2).sum() == 0:
        return 0
    return (np.bitwise_and(v1, v2).sum() / (np.bitwise_or(v1, v2).sum()))

get_fp = lambda mol: AllChem.GetMorganFingerprintAsBitVect(
    mol, radius=2, nBits=512)

def get_qspr(dataset):
    mol_repres, scaff_repres, motif_repres = [], [], []
    for smi in dataset.smiles:
        res = get_multi_repres(smi)
        motif_repres.append(res[0])
        mol_repres.append(res[1])
        scaff_repres.append(res[2])
    labels = np.array(dataset.labels)

    motif_diffs, mol_diffs, scaff_diffs = [], [], []
    for i in range(len(motif_repres)):
        for j in range(i+1, len(motif_repres)):
            motif_diffs.append(1 - tanimoto(motif_repres[i], motif_repres[j]))
    
        mol_diffs += list(1 - np.array(DataStructs.BulkTanimotoSimilarity(mol_repres[i], mol_repres[i+1:])))
        scaff_diffs += list(1 - np.array(DataStructs.BulkTanimotoSimilarity(scaff_repres[i], scaff_repres[i+1:])))

    label_diff = abs(labels - labels.reshape(1, -1))
    label_diff = label_diff[np.triu_indices(label_diff.shape[0], k=1)]

    correlations = np.zeros(3)
    correlations[0] += (np.corrcoef(mol_diffs, label_diff)[1][0])
    correlations[1] += (np.corrcoef(scaff_diffs, label_diff)[1][0])
    correlations[2] += (np.corrcoef(motif_diffs, label_diff)[1][0])
    return correlations / correlations.sum()
```

```python
# Setup model
model = GNNPredictor(num_layer=config['model']['num_layer'],
                     emb_dim=config['model']['emb_dim'],
                     num_tasks=dataset.num_task,
                     normalize=config['model']['normalize'],
                     atom_feat_dim=170,
                     bond_feat_dim=14,
                     drop_ratio=config['model']['dropout_ratio'],
                     attn_drop_ratio=config['model']['attn_dropout_ratio'],
                     temperature=config['model']['temperature'],
                     use_prompt=config['model']['use_prompt'],
                     model_head=config['model']['heads'],
                     layer_norm_out=config['model']['layernorm'], 
                     backbone=config['model']['backbone'])

if config['model']['checkpoint']:
    print('Loading checkpoint from {}'.format(config['model']['checkpoint']))
    model.load_state_dict(torch.load(config['model']['checkpoint'])['wrapper'], strict=False)
model.to(config['device'])
```

```python
data_name_list = ['CHEMBL219_Ki', 'CHEMBL231_Ki', 'CHEMBL228_Ki', 'CHEMBL233_Ki', 'CHEMBL214_Ki',
                  'CHEMBL237_EC50', 'CHEMBL2971_Ki', 'CHEMBL4616_EC50', 'CHEMBL235_EC50',
                  'CHEMBL239_EC50', 'CHEMBL4203_Ki', 'CHEMBL3979_EC50', 'CHEMBL2047_EC50',
                  'CHEMBL4005_Ki', 'CHEMBL4792_Ki', 'CHEMBL218_EC50', 'CHEMBL237_Ki', 
                  'CHEMBL1862_Ki', 'CHEMBL204_Ki', 'CHEMBL238_Ki', 'CHEMBL234_Ki',
                  'CHEMBL264_Ki', 'CHEMBL236_Ki', 'CHEMBL2835_Ki', 'CHEMBL244_Ki', 'CHEMBL2147_Ki',
                  'CHEMBL262_Ki', 'CHEMBL1871_Ki', 'CHEMBL287_Ki', 'CHEMBL2034_Ki']
print(len(data_name_list))
```

```python
qspr_list = []
pws_list = []
for data_name in data_name_list:
    with open('./config/moleculeace/chembl.yaml', 'r') as f:
            config = yaml.load(f, Loader=yaml.FullLoader)
    config['dataset']['data_name'] = data_name
    
    # Setup dataset and dataloader
    dataset, train_loader, val_loader, test_loader = get_dataloader(config)
    
    # Get QSPR correlation
    qspr = get_qspr(dataset)
    qspr_list.append(qspr)

    # Initialize prompt:
    pws = []
    for _ in range(5):
        best_initialization = optimize_prompt_weight_ri(model, train_loader, val_loader, config)
        pw = torch.softmax(best_initialization, dim=0).numpy()
        pws.append(pw)
    pws_list.append(pws)
```

```python
# plt.figure(figsize=(12, 8))
# plt.scatter(qspr.flatten(), pw.flatten(), s=180, alpha=0.8, color='dodgerblue', edgecolor='w', linewidth=1)

# xs = np.arange(-0.05, 0.65, 0.01)
# plt.plot(xs, np.poly1d(np.polyfit(qspr.flatten(), pw.flatten(), 1))(xs), 
#          color='dodgerblue', linestyle='-', label='fitted line')

# plt.xlabel('QSPR')
# plt.ylabel('PW')
# plt.show()

def entropy(prob_vector):
    """Calculate the entropy of a probability vector."""
    return -np.sum(prob_vector * np.log(prob_vector + 1e-9))

def apply_temperature_scaling(prob_vector, T):
    """Apply temperature scaling to a probability vector."""
    scaled_probs = np.power(prob_vector, 1.0 / T)
    scaled_probs /= np.sum(scaled_probs)
    return scaled_probs

def objective_function(T, target_entropy, prob_vector):
    """Objective function to minimize the difference between entropies."""
    scaled_probs = apply_temperature_scaling(prob_vector, T)
    current_entropy = entropy(scaled_probs)
    return np.abs(current_entropy - target_entropy)

scaled_qspr = []
for i in range(len(pw)):
    prob_vector1, prob_vector2 = pw[i], qspr[i]

    # Calculate initial entropy for prob_vector1
    initial_entropy1 = entropy(prob_vector1)
    
    # Use optimization to find the temperature for prob_vector2
    result = minimize(objective_function, x0=1.0, args=(initial_entropy1, prob_vector2), bounds=[(0.01, 10)])
    
    # Get the optimal temperature
    optimal_temperature = result.x[0]
    
    # Scale both vectors using their respective temperatures
    scaled_qspr_i = apply_temperature_scaling(prob_vector2, optimal_temperature)
    scaled_qspr.append(scaled_qspr_i)
```

```python
plt.figure(figsize=(12, 8))

combined_data = np.vstack((pw, qspr))

# Perform PCA
pca = PCA(n_components=1)
transformed_data = pca.fit_transform(combined_data[:, :3])

# Split the transformed data back into the two sets
transformed_pw = transformed_data[:len(pw)]
transformed_qspr = transformed_data[len(pw):]


plt.scatter(transformed_pw[:, 0], transformed_qspr[:, 0], s=180, alpha=0.8, color='dodgerblue', edgecolor='w', linewidth=1)

xs = np.arange(-1, 1, 0.01)
plt.plot(xs, np.poly1d(np.polyfit(transformed_pw[:, 0], transformed_qspr[:, 0], 1))(xs), 
         color='dodgerblue', linestyle='-', label='fitted line')

min_val = min(transformed_pw[:, 0].min(), transformed_qspr[:, 0].min())
max_val = max(transformed_pw[:, 0].max(), transformed_qspr[:, 0].max())
plt.plot([min_val-1, max_val+1], [min_val-1, max_val+1], color='red', linestyle='-', label='perfectly aligned (y=x)')

plt.legend(fontsize=16)
# plt.xlim(-0.75, 0.85)
# plt.ylim(-0.22, 0.15)
plt.xlim(-0.5, 0.5)
plt.ylim(-0.5, 0.5)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.xlabel('First component of optimized prompt weights', fontsize=20)
plt.ylabel('First component of QSPR metric', fontsize=20)
plt.title('Relationship between optimized prompt weights and QSPR metric', fontsize=20)
plt.savefig('./pw_versus_qspr.png', dpi=300, bbox_inches='tight')
plt.show()
```
