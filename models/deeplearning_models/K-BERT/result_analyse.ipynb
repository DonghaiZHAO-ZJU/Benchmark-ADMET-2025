```python
import pandas as pd
import numpy as np

all_tasks = ["BBBP","hERG","Mutagenicity","oral_bioavailability","HLM_metabolic_stability","Caco2","HalfLife","VDss"]
classification_tasks = ["BBBP","hERG","Mutagenicity","oral_bioavailability","HLM_metabolic_stability"]
regression_tasks = ["Caco2","HalfLife","VDss"]
split_methods=["random", "scaffold", "Perimeter", "Maximum_Dissimilarity"]

for task in all_tasks:
    different_splits_train_results, different_splits_valid_results, different_splits_test_results = [], [], []
    for split_method in split_methods:
        different_seed_train_results, different_seed_valid_results, different_seed_test_results = [], [], []
        for i in range(5):
            seed=2024+i*10
            result_pd = pd.read_csv(f'./result_from_server/K_BERT_{task}_{split_method}_{seed}_all_result.csv')
            result1 = result_pd[['train_1', 'train_2', 'train_3', 'train_4', 'train_5']].mean(axis=1)
            result2 = result_pd[['val_1', 'val_2', 'val_3', 'val_4', 'val_5']].mean(axis=1)
            result3 = result_pd[['test_1', 'test_2', 'test_3', 'test_4', 'test_5']].mean(axis=1)
            different_seed_train_results.append(result1[0])
            different_seed_valid_results.append(result2[0])
            different_seed_test_results.append(result3[0])
        different_splits_test_results.append(np.mean(different_seed_test_results).round(4))
        if task=='BBBP':
            print(different_seed_test_results)
    print(different_splits_test_results)
            

```
